```python
# Class of array acts as a list
import ctypes 
```

```python
# Task 1: Create list

class my_list:
    def __init__(self):
        self.size = 1
        self.n = 0  
        self.A = self.__make_array(self.size) 
    
    # when an object is in length function 
    def __len__(self):
        return self.n

    def append(self, item):
        if self.n == self.size:
            # resize and copy the old array elements
            self.__resize(self.size*2)
    
         # if space is empty
        else:
            self.A[self.n] = item  # stored in nth position
            self.n += 1

    def __resize(self, new_capacity):
        
        # part 1 - creates a new array with new capacity
        B = self.__make_array(new_capacity)
        self.size = new_capacity

        # part 2 - copy the content of A to B
        for i in range(self.n):
            B[i] = self.A[i]

        # reassign A
        self.A = B 
        # further we have to work on B array but the logic is write in A

    def __make_array(self, capacity):
        # creates a C type array(static, referencetial) with capacity=size
        return (capacity*ctypes.py_object)()
                  
```

```python
# Task 2 - len feature

L = my_list()
print(len(L))
```

```python
# append feature

L.append("Hello")
L.append(3.14)
L.append(True)
L.append(100)
```
