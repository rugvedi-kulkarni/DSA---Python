Creation of linked list

```python
class Node:
    def __init__(self, value):
        self.data = value 
        self.next = None 
 
```

```python
class LinkedList:
    def __init__(self):
        # Empty linked list
        self.head = None
        self.n = 0
        # no of nodes in LL
        # it defines that how may nodes are in a linked list

    # logic for length function= 
    def __len__(self):
        return self.n

    def insert_head(self, value):
        # new node
        new_Node = Node(value)

        # create connection
        new_Node.next = self.head

        # reassign head
        self.head = new_Node

        # increment n
        self.n += 1

    def __str__(self):

        curr = self.head
        res = ""
        while curr != None:
            res += str(curr.data) + "->"
            curr = curr.next

        return res.strip()[:-2]           

    def insert_tail(self, value):
        new_node = Node(value)

        # if we want to insert in an empty linked list
        if self.head == None:
            self.head = new_node
            self.n += 1
            return
            
        curr = self.head

        while curr.next != None:
            curr = curr.next

        # u r at the last node
        curr.next = new_node
        self.n += 1

    def middle_element(self):
        mid = self.__str__()

        # version 1: only for numbers < 10
        # res = ""
        # for i in range(len(mid)):
                
        #         if mid[i] == "-" or mid[i] == ">":
        #             continue
        #         else:
        #             res = res + str(mid[i])

        # new_size = len(res)
        # middle = new_size // 2 + 1
        # print(res[middle - 1])   

        # version 2: for all digits
        res = mid.strip().split("->")
        new_size = len(res)
        middle = (new_size // 2) + 1
        print(res[middle - 1])

    def reverse_list(self):
        mid = self.__str__()
        res = mid.strip().split("->")

        out = ""
        for i in range(len(res))[::-1]:
            out += str(res[int(i)]) + "->"

        return out[:-2]  
        
```

Create a linked list object

```python
L = LinkedList()
```

Length of linked list

```python
print(len(L))
```

Insert from head

```python
L.insert_head(1)
L.insert_head(2)
L.insert_head(8)
L.insert_head(10)
L.insert_head(20)
```

Print linked list

```python
print(L)
```

Insert at end

```python
L.insert_tail(5)
L.insert_tail(9)
```

```python
print(L)
```

Print the middle element

```python
L.middle_element()
```

Linked List before reverse vs after reverse

```python
print(f'Before reverse: {L}')
print(f'After reverse: {L.reverse_list()}')
```
