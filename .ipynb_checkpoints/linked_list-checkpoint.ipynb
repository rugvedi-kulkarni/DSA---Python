Creation of linked list

```python
class Node:
    def __init__(self, value):
        self.data = value 
        self.next = None 
 
```

```python
class LinkedList:
    def __init__(self):
        # Empty linked list
        self.head = None
        self.n = 0
        # no of nodes in LL
        # it defines that how may nodes are in a linked list

    # logic for length function= 
    def __len__(self):
        return self.n

    def insert_head(self, value):
        # new node
        new_Node = Node(value)

        # create connection
        new_Node.next = self.head

        # reassign head
        self.head = new_Node

        # increment n
        self.n += 1

    def __str__(self):

        curr = self.head
        res = ""
        while curr != None:
            res += str(curr.data) + "->"
            curr = curr.next

        return res.strip()[:-2]           

    def insert_tail(self, value):
        new_node = Node(value)

        # if we want to insert in an empty linked list
        if self.head == None:
            self.head = new_node
            self.n += 1
            return
            
        curr = self.head

        while curr.next != None:
            curr = curr.next

        # u r at the last node
        curr.next = new_node
        self.n += 1

    def insert_after(self, old_val, value):
        new_node = Node(value)

        curr = self.head
        while curr != None:
            if curr.data == old_val:
                break
            curr = curr.next

        if curr != None:
            # logic
            new_node.next = curr.next
            curr.next = new_node
        else:
            return "Item not found"
            
        # if we enter the node value whih does not exists then loop runs till the last node
        # and the last node tail node  i.e. None, none has no data 
        # so it will give us error
         # case1: loop breaks u found the element to insert new node after it - here the current is not none
        # case 2: loop runs till none means element is not found
        

     

    def clear(self):
        # reset values
        self.head = None
        self.n = 0

    def delete_head(self):
        # if LL is empty
        if self.head == None:
            return "Empty Linked List"
            
        self.head = self.head.next
        self.n -= 1
        
    def delete_Tail(self):

        if self.head == None:
            return "Empty linked list"
            
        curr = self.head

        # if  ll contains 1 node then;
        if curr.next == None:
            # 1 item is head,  means we have to delete head
            return self.delete_head()

        while curr.next.next != None:
            curr = curr.next

        # curr  -> 2nd last node
        curr.next = None
        self.n -= 1

    def delete_value(self, value):

        if self.head == None:
            return "Empty LL"

        if self.head.data == value:
            # u want to remove the head node
            return self.delete_head()
            
        curr = self.head
            
        while curr.next != None:
            if curr.next.data == value:
                break
            curr = curr.next
            
        # case 1: value found

        if curr.next == None:
            return "Value not found"
        else:
            curr.next = curr.next.next
            
        # case 2: value not found
        
              

    def search_value(self, key):

        # version 1:
        # res = self.__str__()
        # out = res.strip().split("->")
        # for i in range(len(out)):
        #     if out[i] == str(key):
        #         return i

        # return -1

        # version 2:
        curr = self.head
        if self.head == None:
            return "Empty linked list"
            
        idx = 0
        while curr != None:
            if curr.data == key:
                return idx
            idx += 1
            curr = curr.next

        return "Value not exists"

    def search_index(self, idx):

        # version 1
        # res = self.__str__() # original list
        # out = res.strip().split("->")
        # return out[int(idx)]

        # version 2:
        curr = self.head
        
        pos = 0
        while curr != None:
            if pos == idx:
                return curr.data
            curr = curr.next
            pos += 1

        return "IndexError: List index out of range"
            
    def middle_element(self):
        mid = self.__str__()

        # version 1: only for numbers < 10
        # res = ""
        # for i in range(len(mid)):
                
        #     if mid[i] == "-" or mid[i] == ">":
        #         continue
        #     else:
        #         res = res + str(mid[i])

        # new_size = len(res)
        # middle = new_size // 2 + 1
        # print(res[middle - 1])   

        # version 2: for all digits
        res = mid.strip().split("->")
        new_size = len(res)
        middle = (new_size // 2) + 1
        print(res[middle - 1])

    def reverse_list(self):

        # version 1: 
        
        # mid = self.__str__()
        # res = mid.strip().split("->")

        # out = ""
        # for i in range(len(res))[::-1]:
        #     out += str(res[i]) + "->"

        # return out[:-2]  

        # version 2: 
        
        curr = self.head

        res = ""
        while curr != None:
            res += str(curr.data) + " "
            curr = curr.next

        new_res = res.split(" ")
        final_ll = new_res[:-1][::-1]

        return '->'.join(final_ll)
        
        
       
```

Create a linked list object

```python
L = LinkedList()
```

Length of linked list

```python
print(f'Initially the length of linked list is {len(L)}')
```

Insert from head

```python
L.insert_head(1)
L.insert_head(2)
L.insert_head(8)
L.insert_head(10)
L.insert_head(20)
```

Print linked list

```python
print(f'After inserting => {L}')
```

Insert at end

```python
L.insert_tail(5)
L.insert_tail(9)
```

```python
print(f'After inserting from tail => {L}')
```

Print the middle element

```python
print(f'The middle element of linked list is {L.middle_element()}')
```

Linked List before reverse vs after reverse

```python
print(f'Before reverse: {L}')
L.reverse_list()
print(L)
```

Search by value 

```python
val = L.search_value(10)
print(f'Index of value  is {val} (Search by value method)')
```

Search by index

```python
# 20->10->8->2->1
L.search_index(8)
print(f'Value at idx is {val} (Search by index method)')
```

Clear linked list

```python
L.clear()
print(f'Returns empty linked list as {L}')  # empty linked list
```

Delete from head

```python
L.delete_head()
print(f'Linked List after deleting from head {L}')
```

```python
# This code works one a single element also for empty linked list
# L.insert_head(1000)

L.insert_after(5, 200)
print(f'Linked List after inserting any element => {L}')
```

Pop - delete from tail

```python
# Pop (delete from tail)
L.delete_Tail()
print(f'Linked list after deletion from tail {L}')
```

remove - delete by value

```python
L.delete_value(5)
print(f'Linked list after deletion by value {L}')
```
