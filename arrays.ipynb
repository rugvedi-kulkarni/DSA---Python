```python
# Task 1: Create list

class my_list:
    def __init__(self):
        self.size = 1
        self.n = 0  
        self.A = self.__make_array(self.size) 
    
    # to calcuate length of an object => length(object)
    def __len__(self): # special method
        return self.n

    # function or method to create an array
    def __make_array(self, capacity):
        # creates a C type array(static, referencetial) with capacity=size
        return (capacity*ctypes.py_object)()

    def append(self, item):
        if self.n == self.size:
            # resize and copy the old array elements
            self.__resize(self.size*2)
    
         # if space is empty
        
        self.A[self.n] = item  # stored in nth position
        self.n += 1

    # dynamically increase the size
    def __resize(self, new_capacity):
        
        # part 1 - creates a new array with new capacity
        B = self.__make_array(new_capacity)
        self.size = new_capacity

        # part 2 - copy the content of A to B
        for i in range(self.n):
            B[i] = self.A[i]

        # reassign A
        self.A = B 
        # further we have to work on B array but the logic is write in A

    # to return string
    def __str__(self): # special method
        result = ''

        for i in range(self.n):
            result = result + str(self.A[i]) + ','

        return '[' + result[:-1] + ']'

    # special method to return index of an array
    def __getitem__(self, index): # special method

        if 0 <= index < self.n:
            return self.A[index]
        else:
            return "IndexError - Index out of range"

    # to delete element in an array
    def pop(self):
        if self.n == 0:
            return "Empty list"

        print(self.A[self.n - 1])
        self.n = self.n - 1

    # clear all elements of an array
    def clear(self):
        self.n = 0
        self.size = 1

    # search element in an array
    def find(self, item):

        for i in range(self.n):
            if self.A[i] == item:
                return i

        return "ValueError: element not found"

    # insert element at a specific position
    def insert(self, pos, item):
        
        if self.n == self.size:
            self.__resize(self.size**2)

        for i in range(self.n, pos, -1):
            self.A[i] = self.A[i - 1]

        self.A[pos] = item
        self.n = self.n +  1

    # delete item by directly position or index
    def __delitem__(self, pos):
        # delete code

        if 0 <= pos < self.n:     
            for i in range(pos, self.n - 1):
                self.A[i] = self.A[i + 1]

            self.n -= 1

    # remove element by directly searching item
    def remove(self, item):
        position = self.find(item)

        if type(position) == int:
            self.__delitem__(position)
        else:
            return position 

    # sum feature
    def sum(self):
        sum_no = 0
        for i in range(self.n):
            sum_no += self.A[i]
        print(sum_no)

    # sorting 
    def sort(self):
        
        for i in range(self.n):
            for j in range(i, self.n):

                # ascending order
                if self.A[i] > self.A[j]:
                    self.A[i], self.A[j] = self.A[j], self.A[i]  

            print(self.A[i])

    # version 1
    def op(self):
        max_no = 0
        min_no = 0
        for i in range(self.n):
            if i == 0:
                min_no = self.A[i]
            elif i == self.n - 1:
                max_no = self.A[i]

        print(min_no)
        print(max_no)
        
        
    

                                 
```
